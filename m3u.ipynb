````python
import os

# PT-BR: Pasta original
# EN: Source folder
# ES: Carpeta de origen
pasta_origem = '/home/user/Músicas/Playlist/'

# PT-BR: Pasta onde os novos arquivos serão salvos
# EN: Destination folder where new files will be saved
# ES: Carpeta de destino donde se guardarán los nuevos archivos
pasta_destino = '/home/user/Músicas/Playlist/Celular/'

# PT-BR: Caminho a ser substituído
# EN: Path to be replaced
# ES: Ruta a ser reemplazada
caminho_antigo = '/home/alphaai/Músicas/'

# PT-BR: Novo caminho que substituirá o antigo
# EN: New path that will replace the old one
# ES: Nueva ruta que reemplazará la anterior
caminho_novo = '/storage/emulated/0/Music/'

# PT-BR: Cria a pasta de destino, se não existir
# EN: Create destination folder if it doesn't exist
# ES: Crea la carpeta de destino si no existe
os.makedirs(pasta_destino, exist_ok=True)

# PT-BR: Percorre todos os arquivos na pasta de origem
# EN: Iterate through all files in the source folder
# ES: Recorre todos los archivos en la carpeta de origen
for arquivo in os.listdir(pasta_origem):
    
    # PT-BR: Verifica se o arquivo é do tipo .m3u
    # EN: Check if the file is an .m3u file
    # ES: Verifica si el archivo es .m3u
    if arquivo.endswith('.m3u'):
        caminho_arquivo_origem = os.path.join(pasta_origem, arquivo)

        # PT-BR: Lê o conteúdo do arquivo M3U
        # EN: Read the content of the M3U file
        # ES: Lee el contenido del archivo M3U
        with open(caminho_arquivo_origem, 'r', encoding='utf-8') as f:
            linhas = f.readlines()

        # PT-BR: Indica se houve modificação
        # EN: Flag to indicate if changes were made
        # ES: Indica si hubo modificaciones
        modificado = False

        # PT-BR: Percorre cada linha do arquivo
        # EN: Iterate through each line of the file
        # ES: Recorre cada línea del archivo
        for i in range(len(linhas)):
            
            # PT-BR: Verifica se o caminho antigo está na linha
            # EN: Check if the old path exists in the line
            # ES: Verifica si la ruta antigua está en la línea
            if caminho_antigo in linhas[i]:
                
                # PT-BR: Substitui o caminho antigo pelo novo
                # EN: Replace old path with new path
                # ES: Reemplaza la ruta antigua por la nueva
                linhas[i] = linhas[i].replace(caminho_antigo, caminho_novo)
                modificado = True

        # PT-BR: Se houve modificação, salva no destino
        # EN: If modified, save to destination
        # ES: Si hubo cambios, guarda en el destino
        if modificado:
            caminho_arquivo_destino = os.path.join(pasta_destino, arquivo)
            with open(caminho_arquivo_destino, 'w', encoding='utf-8') as f:
                f.writelines(linhas)

# PT-BR: Mensagem final de sucesso
# EN: Final success message
# ES: Mensaje final de éxito
print("Arquivos M3U copiados e caminhos atualizados com sucesso!")
````

````python
Aqui vai um texto explicativo simples, em **Português (Brasil)**, ensinando como usar o código e o que ele faz:

---

### 📌 Como usar o script para corrigir playlists M3U

Esse script em Python serve para **corrigir caminhos de arquivos dentro de playlists `.m3u`**, algo muito útil quando você copia músicas do computador para o celular e os caminhos deixam de funcionar.

#### 🧠 O que o código faz

Ele procura dentro dos arquivos `.m3u` por caminhos antigos (do computador) e substitui por novos caminhos (do celular).
Por exemplo:

* Antes: `/home/alphaai/Músicas/minha-musica.mp3`
* Depois: `/storage/emulated/0/Music/minha-musica.mp3`

---

### 📂 O que você precisa configurar

Antes de rodar o script, você deve ajustar 4 caminhos:

#### 1. Pasta de origem

```python
pasta_origem = '/home/user/Músicas/Playlist/'
```

➡️ Aqui ficam suas playlists `.m3u` originais.

#### 2. Pasta de destino

```python
pasta_destino = '/home/user/Músicas/Playlist/Celular/'
```

➡️ Onde serão salvos os arquivos corrigidos.

#### 3. Caminho antigo

```python
caminho_antigo = '/home/alphaai/Músicas/'
```

➡️ Esse é o trecho que o script vai **procurar dentro das playlists**.

#### 4. Caminho novo

```python
caminho_novo = '/storage/emulated/0/Music/'
```

➡️ Esse é o trecho que vai **substituir o caminho antigo**.

---

### 🔍 O que exatamente ele procura?

Dentro de cada arquivo `.m3u`, o script analisa linha por linha e verifica:

* Se a linha contém o `caminho_antigo`
* Se encontrar, ele substitui pelo `caminho_novo`

Ele **não altera o arquivo original**, apenas cria uma nova versão corrigida na pasta de destino.

---

### ▶️ Como executar

1. Salve o código em um arquivo, por exemplo:

   ```
   corrigir_playlist.py
   ```

2. No terminal, execute:

   ```
   python3 corrigir_playlist.py
   ```

3. Após rodar, você verá a mensagem:

   ```
   Arquivos M3U copiados e caminhos atualizados com sucesso!
   ```

---

### 📱 Quando usar isso?

Use esse script quando:

* Transferir músicas do PC para o celular
* As playlists não funcionarem no celular
* Os caminhos estiverem diferentes entre dispositivos

---

### ⚠️ Dicas importantes

* Certifique-se de que os caminhos estão **exatamente iguais** aos usados nas playlists
* Funciona apenas com arquivos `.m3u`
* Se nada for alterado, o arquivo não será copiado

---

Se quiser depois, posso te mostrar como adaptar isso para:

* reconhecer automaticamente caminhos diferentes
* funcionar com `.m3u8`
* ou atualizar tudo sem precisar configurar manualmente 👍
````
